# Kappa Effective Solver Run

This notebook runs the repository's `compute_kappa_eff` solver for:
- `d = 10`
- `P = 1000`
- `N = 128` for both hidden widths
- `chi = 20`

The solver also needs a bare ridge value, so this notebook sets `kappa_bare = 1.0` explicitly. Change that value if you want a different bare ridge.

```python
from pathlib import Path
import sys

project_root = Path.cwd().resolve()
if not (project_root / 'lib').exists():
    if (project_root.parent / 'lib').exists():
        project_root = project_root.parent
    else:
        project_root = Path('/home/akiva/FCNX-Ensembling')

sys.path.insert(0, str(project_root / 'lib'))

from kappa_eff_solver import compute_kappa_eff

print(f'Using project root: {project_root}')
```

```python
d = 45
P = 50
N = 128
chi = 10.0
kappa_bare = 0.1
num_samples = 50

kappa_eff = compute_kappa_eff(
    d=d,
    P=P,
    kappa_bare=kappa_bare,
    n1=N,
    n2=N,
    chi=chi,
    num_samples=num_samples,
    verbose=True,
)

print(f'kappa_bare = {kappa_bare:.6f}')
print(f'kappa_eff  = {kappa_eff:.10f}')
print(f'correction = {kappa_eff - kappa_bare:.10f}')
```

```python
result = {
    'd': d,
    'P': P,
    'N': N,
    'chi': chi,
    'kappa_bare': kappa_bare,
    'kappa_eff': kappa_eff,
}
result
```
